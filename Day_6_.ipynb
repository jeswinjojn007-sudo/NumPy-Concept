```python
num1 = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

num2 = np.array([
    [10, 20, 30]  
])

addition = num1 + num2

print("Original array:")
print(num1)
print(num2)

print("\nAfter broadcasting 5:")
print(addition)
```

Aggregate Functions

```python
data = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Median:", np.median(data))
```

View() and Copy()

```python
a = np.array([1, 2, 3, 4])
a[2] = 300
b= a.copy()
print(a)
print(b)
```

```python
a = np.array([1, 2, 3, 4])
b = a.copy()
a[2] = 300
print(a)
print(b)

'''Copied array onve the data, any  
Changes made to orginal array  doesn't
impact copy array'''
```

View()

```python
a = np.array([1,2,3,4])
b = a.view()
a[3] = 400
print(a)
print(b)
'''View doesn't own the data so any change
made in the Orginal array doesn't impacts View array'''
```
